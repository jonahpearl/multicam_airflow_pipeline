```python
from pathlib import Path
import cv2
import numpy as np

from multicamera_airflow_pipeline.tim_240731.sync.sync_cameras import CameraSynchronizer
```

```python
def check_if_correct_fps(recording_directory, samplerate):
        # checks if self.samplerate matches the samplerate of videos
        # logger.info("Checking if correct FPS")
        # get the videos in recording directory
        videos = list(recording_directory.glob("*.mp4"))
        if len(videos) == 0:
            raise FileNotFoundError("No videos found in the recording directory.")
        # get the fps of the videos
        fps = []
        for video in videos:
            cap = cv2.VideoCapture(str(video))
            fps.append(cap.get(cv2.CAP_PROP_FPS))
            cap.release()
        # check if fps matches the expected fps, within 1 fps
        if np.all(np.abs(np.array(fps) - samplerate) < 1):
            return True
        else:
            # logger.info(f"Expected FPS: {self.samplerate}, Found FPS: {fps}")
            print(f"Expected FPS: {samplerate}, Found FPS: {fps}")
            return False
```
